```python
from implementation.Cloth import Cloth
from implementation.utils import createRectangularMesh
import numpy as np
np.set_printoptions(threshold=sys.maxsize)
import time
```

```python
# Caida libre
na = 2; nb = 3
# na = 4; nb = 7
np.random.seed(1)
X, T = createRectangularMesh(a = 0.5, b = 0.8, na = na, nb = nb, h = 0.2)
# X[:,2] += 0.7; 
# X += 0.0001*np.random.randn(X.shape[0],3) 
```

```python
self = Cloth(X, T); 
dt = self.estimateTimeStep(L=0.8)
```

```python
# self.plotMesh()
self.setSimulatorParameters(dt = dt, thck = 0.95, mu_s = 0.3, tol = 0.0075, kappa=0.25*1e-4)
```

```python
tf = int(6/dt)
inds = [0]
u = self.positions[inds] # position of corner 0, pin the cloth there
# self.positions[3]
```

```python
self.positions[[0, na-1]].mean(axis=0)
```

```python
# self.plotMesh()
```

```python
start_time = time.time()
for i in range(tf):
    if i == int(tf/2): # For the second half, release the cloth
        inds = []
        u = self.positions[inds]
    self.simulate(u = u, control = inds)
# what simulate method does:
# Stage 1 — start from current state  Store current positions as phi0.  
# Stage 2 — predict motion  Take the unconstrained implicit physics step.  
# Stage 3 — correct that motion  Iteratively enforce:  shear constraints  
#           stretch constraints self-collision constraints.  
# Stage 4 — enforce external contact  Project onto the floor if needed.  
# Stage 5 — commit the new state  Update positions, velocities, and history.
```

```python
print('Time:',time.time()-start_time)
print('Average iterations',self.total_iters/(len(self.history_pos)-1))
```
